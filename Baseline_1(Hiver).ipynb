```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


# ============================================================
# LOAD DATA
# ============================================================

df = pd.read_csv("/content/spotify_customer_classifiedtwice.csv")

df = df.dropna(subset=["customer_tweet", "spotify_reply"])

print("Total examples:", len(df))


# ============================================================
# CHARACTER-LEVEL TF-IDF
# ============================================================

vectorizer = TfidfVectorizer(
    analyzer="char",
    ngram_range=(3, 5)
)

customer_vectors = vectorizer.fit_transform(
    df["customer_tweet"].astype(str)
)


# ============================================================
# FIND MOST SIMILAR CUSTOMER TWEET
# ============================================================

def find_reply(customer_tweet):

    query_vector = vectorizer.transform([customer_tweet])

    similarities = cosine_similarity(
        query_vector,
        customer_vectors
    )[0]

    best_index = similarities.argmax()

    best_similarity = similarities[best_index]

    return (
        df.iloc[best_index]["spotify_reply"],
        df.iloc[best_index]["customer_tweet"],
        df.iloc[best_index]["intent"],
        best_similarity
    )


# ============================================================
# TEST
# ============================================================

test_tweet = "doesn t work and i even tried "

reply, matched_customer, intent, similarity = find_reply(test_tweet)

print("\nNEW CUSTOMER TWEET:")
print(test_tweet)

print("\nMOST SIMILAR HISTORICAL CUSTOMER TWEET:")
print(matched_customer)

print("\nHISTORICAL SPOTIFY REPLY:")
print(reply)

print("\nINTENT:")
print(intent)

print("\nSIMILARITY:")
print(round(similarity, 4))
```
